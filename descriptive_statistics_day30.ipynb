```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df = iris.frame.drop(columns=["target"])
df.columns = ["sepal_length", "sepal_width", "petal_length", "petal_width"]

print(df.shape)
df.head()
```

## Quick summary with describe()

```python
df.describe().round(3)
```

## Central tendency - mean, median, mode

```python
print("Mean")
print(df.mean().round(3))
```

```python
print("Median")
print(df.median())
```

```python
# mode can return more than one value when there is a tie
for col in df.columns:
    print(col, "->", df[col].mode().tolist())
```

```python
# how many times does the most common sepal length appear?
df["sepal_length"].value_counts().head(3)
```

## Dispersion - range, variance, standard deviation, IQR

```python
print("Range (max - min)")
print((df.max() - df.min()).round(2))
```

```python
print("Variance")
print(df.var().round(3))
```

```python
print("Standard deviation")
print(df.std().round(3))
```

```python
q1 = df.quantile(0.25)
q3 = df.quantile(0.75)
iqr = q3 - q1
print("Interquartile range (Q3 - Q1)")
print(iqr.round(2))
```

```python
# coefficient of variation: standard deviation as a % of the mean,
# which makes spread comparable across variables measured on different scales
cv = (df.std() / df.mean() * 100).round(1)
print("Coefficient of variation (%)")
print(cv)
```

## Quartiles and percentiles

```python
df.quantile([0.25, 0.50, 0.75])
```

```python
# 10th and 90th percentiles
df.quantile([0.10, 0.90])
```

```python
# the same percentile calculated with NumPy
print(np.percentile(df["petal_length"], [10, 50, 90]))
```

```python
# percentile rank: what share of flowers have a sepal length of 6.0 cm or less?
share = (df["sepal_length"] <= 6.0).mean() * 100
print(round(share, 1), "%")
```

## A note on standard deviation in Pandas vs NumPy

```python
col = df["sepal_length"]
print("Pandas .std()        :", round(col.std(), 4))
print("NumPy  np.std()      :", round(np.std(col), 4))
print("NumPy  ddof=1        :", round(np.std(col, ddof=1), 4))
```

Pandas divides by n-1 (sample standard deviation) by default, while NumPy divides by n (population standard deviation). Passing `ddof=1` to NumPy makes the two match.

## All measures in one table

```python
summary = pd.DataFrame({
    "mean": df.mean(),
    "median": df.median(),
    "mode": df.mode().iloc[0],
    "min": df.min(),
    "max": df.max(),
    "range": df.max() - df.min(),
    "variance": df.var(),
    "std": df.std(),
    "Q1": q1,
    "Q3": q3,
    "IQR": iqr,
    "CV %": cv,
}).round(2)

summary
```

## Checking mean against median

```python
gap = (df.mean() - df.median()).round(3)
print("mean - median")
print(gap)
```

```python
# petal length has a large gap, so look at how its values are distributed
bins = [1, 2, 3, 4, 5, 6, 7]
print(pd.cut(df["petal_length"], bins=bins, right=False).value_counts().sort_index())
```

## Outliers using the IQR rule

```python
for col in df.columns:
    low = q1[col] - 1.5 * iqr[col]
    high = q3[col] + 1.5 * iqr[col]
    outliers = sorted(df[col][(df[col] < low) | (df[col] > high)].tolist())
    print(col, "->", outliers)
```

## Written interpretation

**Central tendency.** Sepal length and sepal width are close to symmetric: their means (5.84 and 3.06) are within 0.06 of their medians (5.80 and 3.00). Petal length is not. Its mean is 3.76 but its median is 4.35, a gap of 0.59, and it has two modes (1.4 and 1.5 are tied). The bin counts explain why: 50 flowers have a petal length below 2 cm, none fall between 2 and 3 cm, and the other 100 are at 3 cm or above. That is the setosa species forming its own group, so the mean of 3.76 sits between the two groups and is not a typical value for either.

**Dispersion.** The raw standard deviation makes petal length (1.77) look much more variable than sepal length (0.83), but the variables are on different scales. The coefficient of variation puts them on equal footing: petal width varies the most relative to its size (63.6%), then petal length (47.0%), while sepal width (14.3%) and sepal length (14.2%) are far more consistent. The IQR tells the same story, with 3.5 cm for petal length against 1.3 cm for sepal length.

**Quartiles and percentiles.** The middle half of sepal lengths falls between 5.1 and 6.4 cm. For petal length, the 10th percentile is 1.4 cm and the 90th is 5.8 cm, so only 10% of flowers have longer petals than 5.8 cm. About 59% of flowers have a sepal length of 6.0 cm or less, which puts 6.0 cm near the 59th percentile.

**Outliers.** Using the 1.5 x IQR rule, sepal width is the only variable with outliers: 2.0, 4.1, 4.2 and 4.4. Sepal length, petal length and petal width have none.

## Interview questions

**What is the difference between mean, median, and mode?**
The mean is the sum of all values divided by the count, so extreme values pull it up or down. The median is the middle value once the data is sorted, so it is not affected by how extreme the outer values are. The mode is the most frequent value, and there can be more than one, as with petal length above. When the mean and median are far apart, as in petal length (3.76 vs 4.35), the data is usually skewed or made up of more than one group.

**What does standard deviation tell us?**
How far values typically sit from the mean, in the same units as the data. A small standard deviation means values cluster tightly around the average, a large one means they are spread out. It is hard to compare across variables on different scales, so dividing by the mean to get the coefficient of variation helps, as shown above.

**What is a percentile?**
The value below which a given percentage of the data falls. The 90th percentile of petal length is 5.8 cm, meaning 90% of flowers have a petal length of 5.8 cm or less. The 25th, 50th and 75th percentiles are the quartiles, and the 50th percentile is the median.
